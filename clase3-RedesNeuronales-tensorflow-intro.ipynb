Importar Librerías

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

Creación de datos

```python
def funcionPrueba(x):
    return 0.5*x**3 - 2*x**2 + x + 3

x = np.linspace(-3,5,40) 
y = funcionPrueba(x)

print(x.shape)

plt.figure()
plt.plot(x,y)
plt.title("Funcion cubica original")
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

Definición de la data de Entrenamiento

```python
x_train = x.reshape(-1,1)
y_train = y.reshape(-1,1)

print(f"El shape de x_train es :{x_train.shape}")
print(f"El shape de y_train es :{y_train.shape}")
```

- El shape de un array unidimensional es (elementos)

Definición del diseño del modelo

```python
learning_rate = 0.01 # Probar con 0.001 y 0.1

modelo = tf.keras.Sequential([
    tf.keras.layers.Dense(16, activation='relu', input_shape=(1,)), #input_shape permite definir la dimensión del vector de entrada
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dense(1)
])
```

Diseño: 
- 16 significa que la capa oculta tiene 16 neuronas
- 32 significa que la capa oculta tiene 32 neuronas
- 1 significa que la capa de salida tiene 1 neurona
- input_shape=(1,) significa que la capa de entrada tiene 1 neurona, que se refiere a la dimensión del vector de entrada
- Para este caso (40, 1) significa que el vector de entrada tiene 40 filas y 1 columna, esta columna indica que el vector de entrada es unidimensional  
- Si tuviera (40,4) significa que el vector de entrada tiene 40 filas y 4 columnas, las columnas indican que tengo 4 neuronas en la capa de entrada

La función de activación:

- En redes neuronales permite introducir no linealidad en el modelo, lo que posibilita aprender patrones complejos y relaciones no lineales en los datos. Además, determina si una neurona debe activarse, limita los valores de salida y facilita el entrenamiento mediante la retropropagación de gradientes
- La función de activación es la función que se utiliza para transformar el vector de entrada, esta función es una función matemática que se aplica a cada neurona de la capa oculta para 
- La activation='relu' significa que la función de activación es la función ReLU. Aplicada en su mayoría para las capas ocultas.
- La activation='sigmoid' significa que la función de activación es la función sigmoide
- La activation='tanh' significa que la función de activación es la función tangente hiperbólica
- En la capa de salida, no se utiliza una función de activación, ya que la función de activación es la función lineal
- La función lineal es la función que no realiza ninguna transformación al vector de entrada, es decir, el vector de entrada sale igual que el vector de entrada

```python
optimizador = tf.keras.optimizers.Adam(learning_rate=learning_rate)

modelo.compile(
    optimizer=optimizador,
    loss='mse' # Función de costo porque es una regresión
)
```

Compile:
- El compile permite configurar el modelo para el entrenamiento
- Compilar un modelo en TensorFlow con Keras se realiza usando model.compile() para configurar el proceso de entrenamiento, definiendo el optimizador, la función de pérdida y las métricas. Es esencial antes de .fit(), especificando cómo el modelo calculará el error y mejorará.

Learning Rate:
- El learning rate (tasa de aprendizaje) es un hiperparámetro crítico que controla cuánto ajusta un modelo sus pesos internos en cada iteración al minimizar el error.
- Generalmente entre 0.0 y 1.0, determina la velocidad de convergencia: un valor alto acelera el aprendizaje pero puede volverse inestable, mientras que uno bajo es preciso pero lento.

Función de optimización:
- Se define un optimizador, que es el algoritmo que se va a utilizar para ajustar los pesos de las neuronas. Estos están basados en el descenso del gradiente
- El descenso del gradiente es un algoritmo que se utiliza para ajustar los pesos de las neuronas, este algoritmo se basa en la derivada de la función de costo
- La derivada de la función de costo es la función que mide la diferencia entre el valor real y el valor predicho

Función de costo:
- Función de Costo vs. Pérdida: La función de pérdida mide el error en un solo punto de datos, mientras que la función de coste calcula el promedio de todas las pérdidas en el conjunto de entrenamiento.
- Función de Costo en Regresión: El Error Cuadrático Medio (MSE - Mean Squared Error) calcula la distancia al cuadrado entre el valor predicho y el real, promediándolos.
- Función de Costo en Clasificación: Se utilizan funciones como la pérdida logarítmica (Log Loss) o la entropía cruzada para medir la diferencia entre las probabilidades predichas y la clase real.
- Minimización: El algoritmo de descenso de gradiente ajusta los parámetros del modelo (pesos y sesgo) en la dirección que reduce el valor de la función de costo.
- Importancia: Un costo pequeño indica un buen ajuste del modelo, mientras que un costo alto indica un rendimiento pobre.

```python
# Permite revisar el resumen del modelo
modelo.summary() 
```

```python
historia = modelo.fit(
    x_train,
    y_train,
    epochs = 800, # cantidad de iteraciones que se van a realizar para el entrenamiento
    # batch_size = x_train.shape[0],
    batch_size = 16, # cantidad de datos que se van a utilizar para el entrenamiento
    verbose=1
)
```

Diferencia entre Epochs y Batch Size:

- Epochs: cantidad de iteraciones que se van a realizar para el entrenamiento
- Batch size: cantidad de datos que se van a utilizar para el entrenamiento. Se calcula dividiendo el total de datos entre el número de épocas.

Ejemplo:
- Si tengo 10000 ejemplos
- Epochs: indica el número de iteraciones
- Si el batch size es de 256, se divide internamente 10000/256 entonces se tendrá 39.06 subconjuntos de datos (+ 1 incompleto).
- Internamente se tendrán 39 * 100 actualizaciones de pesos

Otros features:
- verbose: cantidad de información que se va a mostrar durante el entrenamiento
- loss: función de costo que se va a utilizar para el entrenamiento
- optimizer: algoritmo que se va a utilizar para el entrenamiento
- metrics: métricas que se van a utilizar para el entrenamiento
